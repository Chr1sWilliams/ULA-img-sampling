```python
import numpy as np
import torch
from PIL import Image
import blobfile as bf

def load_image_like_dataset(path, resolution):
    with bf.BlobFile(path, "rb") as f:
        pil_image = Image.open(f)
        pil_image.load()

    pil_image = pil_image.convert("L")  # same as dataset

    # Match the dataset logic, but avoid zero-sized intermediate images.
    while min(pil_image.size) >= 2 * resolution:
        new_size = tuple(max(1, x // 2) for x in pil_image.size)
        pil_image = pil_image.resize(new_size, resample=Image.BOX)

    scale = resolution / min(pil_image.size)
    new_size = tuple(max(1, round(x * scale)) for x in pil_image.size)
    pil_image = pil_image.resize(new_size, resample=Image.BICUBIC)

    arr = np.array(pil_image.convert("RGB"))
    crop_y = (arr.shape[0] - resolution) // 2
    crop_x = (arr.shape[1] - resolution) // 2
    arr = arr[crop_y:crop_y + resolution, crop_x:crop_x + resolution]
    arr = arr.astype(np.float32) / 127.5 - 1

    x = np.transpose(arr, [2, 0, 1])   # (3, H, W)
    x = torch.from_numpy(x).unsqueeze(0)  # (1, 3, H, W)
    to_gray = transforms.Grayscale(num_output_channels=1)
    x = to_gray(x)

    return x
```

```python
base_uv = "bh_util/LLDataChecks/simdata_3598_grmhd{g}_seed3.uvfits"
base_img = "bh_util/LLDataChecks/simdata_3597_grmhd{g}_seed3.png"

import contextlib
import os
import warnings
warnings.filterwarnings("ignore")


def orientations_4d(x):
    variants = []
    for k in range(4):
        r = torch.rot90(x, k, dims=(2, 3))
        variants.append((f"rot{k*90}", r))
        variants.append((f"rot{k*90}_flipH", torch.flip(r, dims=(3,))))
    return variants

for g in range(1, 6):
    uvfile = base_uv.format(g=g)
    x_file = base_img.format(g=g)

    with open(os.devnull, 'w') as f, contextlib.redirect_stdout(f):
        x = load_image_like_dataset(x_file, 128)

    x = torch.clamp(x + 1, min=1e-15)

    # Remove batch and channel dimensions
    img = x.squeeze()  # shape -> (128, 128)

    # Convert to numpy (if needed)
    img = img.detach().cpu().numpy()

    # Plot
    plt.imshow(img, cmap='afmhot', vmin = 0)
    plt.colorbar()
    plt.title("Tensor Image")
    plt.show()

    with open(os.devnull, 'w') as f, contextlib.redirect_stdout(f):
        cd = conditional_diffusion.DiffusionGuidance(device, uvfile, psize)

    print(f"\nProcessing GRMHD{g}")
    print("uvfile:", uvfile)
    print("x_file:", x_file)

    results = []
    for name, x_var in orientations_4d(x):
        ll = cd.log_likelihood(x_var, cd.A_fwd)
        ll_val = ll.item() if torch.is_tensor(ll) else ll
        results.append((name, ll_val))
        print(f"{name}: {ll_val}")

    best_name, best_ll = max(results, key=lambda t: t[1])
    print("Best:")
    print(best_name, best_ll)
```
